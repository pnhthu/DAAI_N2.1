```python
import pandas as pd

# 1. Đọc file dữ liệu sản phẩm gốc
df_product = pd.read_csv('D:\products.csv')
print(f'Số dòng ban đầu: {len(df_product)}')

# 2. Xóa các dòng bị trùng lặp hoàn toàn
df_product = df_product.drop_duplicates()

# 3. Làm sạch chuỗi văn bản (strip khoảng trắng thừa, chuẩn hóa chữ hoa/thường nếu cần)
# Giả sử bảng có các cột dạng text như 'product_name', 'category', 'segment'...
# Bạn điều chỉnh tên cột cho đúng với file products.csv của bạn nhé:
text_cols = df_product.select_dtypes(include=['object']).columns
for col in text_cols:
  df_product[col] = df_product[col].astype(str).str.strip()

# 4. Xử lý giá trị khuyết thiếu (Missing values)
# Ví dụ: Nếu thiếu giá trị ở mã sản phẩm (product_id) thì xóa dòng đó, thiếu giá số thì điền 0 hoặc median/mean
if 'product_id' in df_product.columns:
  df_product = df_product.dropna(subset=['product_id'])

# 5. Chuyển đổi kiểu dữ liệu cho đúng chuẩn (ví dụ cột giá 'price' hoặc 'cogs' sang float)
if 'price' in df_product.columns:
  df_product['price'] = pd.to_numeric(
      df_product['price'].astype(str).str.replace(',', ''), errors='coerce'
  )

# 6. Chuẩn hóa tên cột về dạng snake_case (chữ thường, dùng dấu gạch dưới)
df_product.columns = (
    df_product.columns.str.strip().str.lower().str.replace(' ', '_')
)

# 7. Xuất ra file product.csv đã làm sạch chuẩn
df_product.to_csv('product.csv', index=False, encoding='utf-8-sig')

print(f'✅ Đã làm sạch và tạo xong file: product.csv (còn {len(df_product)} dòng)')
print(df_product.head())
```
