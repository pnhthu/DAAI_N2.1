```python
import pandas as pd

# Đường dẫn file nguồn và đầu ra
geo_file_path      = '/content/geography.csv'
csv_geography_path = '/content/GEOGRAPHY.csv'

try:
    raw_geo = pd.read_csv(geo_file_path)

    # Tạo lại bảng danh mục district để lấy chính xác mã district_id
    district_pairs = raw_geo[['district', 'city']].drop_duplicates().dropna().reset_index(drop=True)
    district_pairs['district_id'] = [f'DIS{str(i+1).zfill(5)}' for i in range(len(district_pairs))]

    # Từ điển tra cứu cặp (district, city) -> district_id
    pair_to_dis_id = dict(zip(zip(district_pairs['district'], district_pairs['city']), district_pairs['district_id']))

    # Ánh xạ zip sang district_id
    geography_df = raw_geo[['zip', 'district', 'city']].drop_duplicates(subset=['zip']).reset_index(drop=True)
    geography_df['district_id'] = geography_df.apply(lambda row: pair_to_dis_id.get((row['district'], row['city'])), axis=1)

    # Lấy 2 thuộc tính theo Schema (zip, district_id)
    geography_df = geography_df[['zip', 'district_id']]
    geography_df.to_csv(csv_geography_path, index=False, encoding='utf-8-sig')
    print('Da tao file GEOGRAPHY.csv thanh cong.')

except FileNotFoundError:
    print(f'Loi: Khong tim thay file {geo_file_path}.')
```
