```python
import pandas as pd
from sqlalchemy import create_engine
from google.colab import files
```

```python

print("Đang đọc dữ liệu đơn hàng...")
raw_orders = pd.read_excel('orders_enriched(1).csv')

raw_orders['order_date'] = pd.to_datetime(raw_orders['order_date'], unit='D', origin='1899-12-30')

columns_3nf = [
    'order_id', 'order_date', 'order_status', 'device_type',
    'order_source', 'customer_id', 'sales_employee_id', 'zip'
]
orders_df_final = raw_orders[columns_3nf].copy()

orders_df_final.to_csv('ORDERS_3NF.csv', index=False)
print("Đã tạo xong file CSV. Đang tải về máy...")
files.download('ORDERS_3NF.csv')
```
