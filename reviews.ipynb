```python
import pandas as pd

# 1. Đọc file dữ liệu đánh giá gốc
df_review = pd.read_csv('D:\reviews.csv')
print(f'Số dòng ban đầu (review): {len(df_review)}')

# 2. Xóa các dòng bị trùng lặp hoàn toàn
df_review = df_review.drop_duplicates()

# 3. Làm sạch chuỗi văn bản (strip khoảng trắng thừa)
text_cols_review = df_review.select_dtypes(include=['object']).columns
for col in text_cols_review:
  df_review[col] = df_review[col].astype(str).str.strip()

# 4. Xử lý giá trị khuyết thiếu (Missing values)
if 'review_id' in df_review.columns:
  df_review = df_review.dropna(subset=['review_id'])

# 5. Chuyển đổi kiểu dữ liệu cho đúng chuẩn (ví dụ: rating sang int, review_date sang datetime)
if 'rating' in df_review.columns:
  df_review['rating'] = pd.to_numeric(
      df_review['rating'], errors='coerce'
  ).fillna(0).astype(int)

if 'review_date' in df_review.columns:
  df_review['review_date'] = pd.to_datetime(df_review['review_date'], errors='coerce')

# 6. Chuẩn hóa tên cột về dạng snake_case
df_review.columns = (
    df_review.columns.str.strip().str.lower().str.replace(' ', '_')
)

# 7. Xuất ra file review.csv đã làm sạch chuẩn
df_review.to_csv('review.csv', index=False, encoding='utf-8-sig')

print(f'✅ Đã làm sạch và tạo xong file: review.csv (còn {len(df_review)} dòng)')
print(df_review.head())
```
