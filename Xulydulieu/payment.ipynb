```python
import pandas as pd
from sqlalchemy import create_engine
from google.colab import files

print("Đang đọc dữ liệu thanh toán và danh sách đơn hàng...")
df_payments = pd.read_csv('payments.csv')

df_orders = pd.read_csv('ORDERS_3NF.csv')

print("Đang xử lý chuẩn hóa 3NF...")

columns_3nf = ['order_id', 'payment_method', 'payment_value', 'installments']
df_payment_final = df_payments[columns_3nf].copy()

valid_order_ids = df_orders['order_id'].unique()

df_payment_final = df_payment_final[df_payment_final['order_id'].isin(valid_order_ids)]

print(f"Số lượng giao dịch hợp lệ sau khi lọc: {len(df_payment_final)}")

filename = 'PAYMENT_3NF.csv'
df_payment_final.to_csv(filename, index=False)
```
