```python
import pandas as pd

# Đường dẫn file nguồn và đầu ra
geo_file_path     = '/content/geography.csv'
csv_district_path = '/content/DISTRICT.csv'

try:
    raw_geo = pd.read_csv(geo_file_path)

    # Ánh xạ city_name -> city_id đồng bộ với bảng CITY
    unique_cities = raw_geo['city'].dropna().unique()
    city_map = {city_name: f'CT{str(i+1).zfill(3)}' for i, city_name in enumerate(unique_cities)}

    # Trích xuất cặp (district, city) duy nhất
    district_pairs = raw_geo[['district', 'city']].drop_duplicates().dropna().reset_index(drop=True)
    district_pairs['district_id']   = [f'DIS{str(i+1).zfill(5)}' for i in range(len(district_pairs))]
    district_pairs['district_name'] = district_pairs['district']
    district_pairs['city_id']       = district_pairs['city'].map(city_map)

    # Lấy 3 thuộc tính theo Schema (district_id, district_name, city_id)
    district_df = district_pairs[['district_id', 'district_name', 'city_id']]
    district_df.to_csv(csv_district_path, index=False, encoding='utf-8-sig')
    print('Da tao file DISTRICT.csv thanh cong.')

except FileNotFoundError:
    print(f'Loi: Khong tim thay file {geo_file_path}.')
```
