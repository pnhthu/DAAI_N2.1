```python
import pandas as pd
from google.colab import files

print("Đang đọc dữ liệu đơn hàng...")

raw_orders = pd.read_excel('orders_enriched(1).csv')

print("Đang xử lý bảng EMPLOYEE...")

employee_cols = [
    'sales_employee_id', 'sales_employee_name',
    'marital_status', 'education_level', 'years_experience'
]
df_emp = raw_orders[employee_cols].drop_duplicates().copy()

raw_orders['order_date'] = pd.to_datetime(raw_orders['order_date'], unit='D', origin='1899-12-30')
anchor_year = raw_orders['order_date'].dt.year.max()

df_emp['hire_date'] = pd.to_datetime((anchor_year - df_emp['years_experience']).astype(str) + '-01-01')

final_columns = [
    'sales_employee_id', 'sales_employee_name',
    'marital_status', 'education_level', 'hire_date'
]
df_emp_final = df_emp[final_columns]

print(f"Tổng số nhân viên độc lập bóc tách được: {len(df_emp_final)}")

df_emp_final.to_csv('EMPLOYEE_3NF.csv', index=False)
```
